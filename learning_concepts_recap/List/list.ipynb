# List comprehension

```python
prev_list=[1,2,3]
new_list=[]

for i in prev_list:
    multiply=i*2
    new_list.append(multiply)



print(new_list)
```

```python
# compre_new_List=[source_item_we_can_perform_operation_here for item_from_the_source_list in source_list]
prev_list=[1,2,3]

compre_new_List=[i*2 for i in prev_list]
print(compre_new_List)



oldword="hari"
newwordlist=[info for info in oldword]
print(newwordlist)
```

# Conditional List Comprehension



```python

# newlist=[new_item for item in list if condition]


pew_list=[-1,-2,1,212,-33,4]


newlist=[number for number in pew_list if number>0]


print(newlist)
```

```python
def twoSum(nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        output=[]
        for i in range(len(nums)):
            for j in range(i+1,len(nums)):
                if nums[i]+nums[j]==target:
                    output.extend([i,j])
        return output
            



numslist=[2,7,11,15]
targetnumber=9

print(twoSum(nums=numslist,target=targetnumber))
```
