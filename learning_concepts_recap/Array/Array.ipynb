# Array traversal

```python
import array



arr1=array.array('i',[1,2,3,4,5,6,7,8])
arr2=array.array('d',[1.2,1.34,23.5,65.3])


print(arr1)
print(arr2)


def traversearray(arr):
    for i in arr: # o (n)
        print(i) # o (1)


#time cmp: o(n)
#space cmp: o(1)

traversearray(arr1)
```

# Searching for Element

## 1. Linear Search algo- Iterate through element in array one by one.

```python
import array 


arr1=array.array('i',[1,2,3,54,6,7])




def linear_search(arr,element):
    for i in arr:
        if element==i:
            print("Element Found")
        else:
            print("Not Found")


linear_search(arr1,3)




def linear_search2(arr,element):
    for i in range(len(arr)):
        if element==arr[i]:
            print("element found")
        else:
            print("element not found")
    
      


linear_search2(arr1,3)


 

def linear_search3(arr,element):
    for i in range(len(arr)):
        if element==arr[i]:
            return i
    return -1
    
    
print(linear_search3(arr1,3))
```

```python
from array import array


my_sample_array=array('i',[1,2,3,4,5])
print(my_sample_array)

# 1 traverse array 
for i in my_sample_array:
    print(i)


# 2 accesa individual elements through indexes

print(my_sample_array[2])


# 3 append value to the array using append method - it will add only in the end of the array

my_sample_array.append(12)
print(my_sample_array)


# 4 insert value in the array using insert method - need index and the value

my_sample_array.insert(1,23)
print(my_sample_array)


#5 extend array using extend() method

my_sample_array_1=array('i',[23,43,54])
my_sample_array.extend(my_sample_array_1)
print(my_sample_array)

# 6 add item from list to array using fromlist() method

mytemplist=[234,234,565]
my_sample_array.fromlist(mytemplist)
print(my_sample_array)


# 7 remove any element from the array using remove() method 
my_sample_array.remove(565)
print(my_sample_array)


# 8 remove last element from the array

my_sample_array.pop()
print(my_sample_array)



# 9 fetch any element using array index

print(my_sample_array.index(12))



# 10 reverse a python array using reverse () method
my_sample_array.reverse()
print(my_sample_array)


# 11 get array buffer information using buffer_info() method

print(my_sample_array.buffer_info())



# 12 to count the occurence of the element 

print(my_sample_array.count(23))



#13 converting array to list

sampplelist=my_sample_array.tolist()
print(sampplelist)



# array slicing

print(my_sample_array[1:6])
```

```python
import numpy as np
twodarray=np.array([
    [23,21,34,56],
    [22,34,32,56],
    [23,55,76,77],
    [22,44,55,66]
])

print(twodarray)




def accesselement(array,rowindex,colindex):
    if rowindex>=len(array) and colindex>=len(array[0]):
        print("incorrect index")
    else:
        print(array[rowindex][colindex])


accesselement(twodarray,1,0)
accesselement(twodarray,1,1)
accesselement(twodarray,1,2)
accesselement(twodarray,1,3)
```

```python
import numpy as np
twodarray=np.array([
    [23,21,34,56],
    [22,34,32,56],
    [23,55,76,77],
    [22,44,55,66]
])

print(twodarray)


print(len(twodarray))
print(len(twodarray[0]))


def traverse2d(array):
    for i in range(len(array)):
        for j in range(len(array[0])):
            print(array[i][j])


traverse2d(twodarray)
```

```python
arr = [1, 7, 3, 4, 9, 5]

def max_product(array):
    max1=0
    max2=0

    for number in array:
        if number>max1:
            max2=max1
            max1=number
        elif number>max2:
            max2=number
    print(max1,max2)
    return (max1-1)*(max2-1)



max_product(arr)#63(9*7)
```

```python
def pair_sum(arr,target):
    temp=[]
    n=len(arr)
    for i in range(n):
        for j in range(i+1,n):
            if arr[i]+arr[j]==target:
                print(arr[i]+arr[j])
                temp.append(f"{arr[i]}+{arr[j]}")
                print(temp)

        
    return temp












print(pair_sum([2, 4, 3, 5, 6, -2, 4, 7, 8, 9],7))

# Output : ['2+5', '4+3', '3+4', '-2+9']
```
