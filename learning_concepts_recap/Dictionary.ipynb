```python
def maxFrequencyElements(nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        dicst={}
        value=0
        for num in nums:
            if num in dicst:
                dicst[num]=dicst[num]+1
            else:
                dicst[num]=1
    
        # maximum=max(dicst.values())
        maximum=0

        for freq in dicst.values():
            if freq>maximum:
                maximum=freq

        

        for number in dicst.values():
            if number==maximum:
                value=value+number

        return value        

nums = [1,2,2,3,1,4]
print(maxFrequencyElements(nums))
```
